```python
import shutil, time
import numpy as np
from pathlib import Path

t0 = time.time()
src = sorted(Path("/kaggle/input").rglob("test_probs_fold4.npy"))
print("sources:", [str(p) for p in src], flush=True)
assert src, "fold4ft probs not mounted"
EXPECTED = 3000 * 4 * 128 * 800 * 2
size = src[0].stat().st_size
print(f"size {size/1e9:.2f}GB expected {EXPECTED/1e9:.2f}GB", flush=True)
assert EXPECTED <= size <= EXPECTED + 4096, f"incomplete probs file: {size} vs data {EXPECTED}"
dst = Path("/kaggle/working/test_probs_ft4.npy")
shutil.copyfile(src[0], dst)
a = np.load(dst, mmap_mode="r")
print(f"copied -> {dst} | shape {a.shape} {a.dtype} | {time.time()-t0:.0f}s", flush=True)
```

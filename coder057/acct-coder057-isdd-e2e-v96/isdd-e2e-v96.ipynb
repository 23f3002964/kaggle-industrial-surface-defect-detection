# E2E variant v96 — per-class targets {'defect_a': 670, 'defect_b': 492, 'defect_c': 1468, 'defect_d': 1180}

End-to-end: 5 checkpoints -> hflip-TTA -> blend -> calibrated masks.


```python

import time, gc
import numpy as np, pandas as pd, cv2
from pathlib import Path
from PIL import Image
from tqdm import tqdm
import torch, torch.nn as nn, torch.nn.functional as F

t_start = time.time()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device, "| GPUs:", torch.cuda.device_count(), flush=True)
ROOT = Path("/kaggle/input/competitions/industrial-surface-defect-detection/public/")
TEST_DIR, SAMPLE = ROOT / "test", ROOT / "sample_submission.csv"
DEFECT_TYPES = ["defect_a", "defect_b", "defect_c", "defect_d"]
TUNED = {"thresholds": [0.175, 0.05, 0.375, 0.2],
         "min_component": [48, 32, 160, 256], "min_pixels": [64, 0, 192, 384]}
BATCH = 64
```

```python

class DoubleConv(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.double_conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels), nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels), nn.ReLU(inplace=True))

    def forward(self, x):
        return self.double_conv(x)


class DownBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.down = nn.Sequential(nn.MaxPool2d(2), DoubleConv(in_channels, out_channels))

    def forward(self, x):
        return self.down(x)


class AttentionGate(nn.Module):
    def __init__(self, gating_channels, skip_channels, inter_channels):
        super().__init__()
        self.W_g = nn.Sequential(nn.Conv2d(gating_channels, inter_channels, 1, bias=False), nn.BatchNorm2d(inter_channels))
        self.W_x = nn.Sequential(nn.Conv2d(skip_channels, inter_channels, 1, bias=False), nn.BatchNorm2d(inter_channels))
        self.psi = nn.Sequential(nn.Conv2d(inter_channels, 1, 1), nn.BatchNorm2d(1), nn.Sigmoid())
        self.relu = nn.ReLU(inplace=True)

    def forward(self, g, x):
        return x * self.psi(self.relu(self.W_g(g) + self.W_x(x)))


class AttentionUpBlock(nn.Module):
    def __init__(self, in_channels, skip_channels, out_channels):
        super().__init__()
        self.up = nn.ConvTranspose2d(in_channels, out_channels, 2, stride=2)
        self.attention = AttentionGate(out_channels, skip_channels, max(out_channels // 2, 1))
        self.conv = DoubleConv(out_channels + skip_channels, out_channels)

    def forward(self, x, skip):
        x = self.up(x)
        if x.shape[-2:] != skip.shape[-2:]:
            x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return self.conv(torch.cat([x, self.attention(x, skip)], dim=1))


class AttentionDefectUNet(nn.Module):
    def __init__(self, in_channels=1, num_classes=4):
        super().__init__()
        self.encoder = nn.Sequential(DoubleConv(in_channels, 32), DownBlock(32, 64),
                                     DownBlock(64, 128), DownBlock(128, 256))
        self.bottleneck = nn.Sequential(nn.MaxPool2d(2), DoubleConv(256, 512))
        self.decoder = nn.ModuleList([AttentionUpBlock(512, 256, 256), AttentionUpBlock(256, 128, 128),
                                      AttentionUpBlock(128, 64, 64), AttentionUpBlock(64, 32, 32)])
        self.outc = nn.Conv2d(32, num_classes, 1)

    def forward(self, x):
        skips = []
        for layer in self.encoder:
            x = layer(x); skips.append(x)
        x = self.bottleneck(x)
        for layer, skip in zip(self.decoder, reversed(skips)):
            x = layer(x, skip)
        return self.outc(x)


def rle_encode(mask):
    pixels = (np.asarray(mask) > 0).flatten(order="F")
    padded = np.concatenate([[0], pixels, [0]])
    changes = np.flatnonzero(padded[1:] != padded[:-1]) + 1
    changes[1::2] -= changes[::2]
    return " ".join(map(str, changes))


def cc_filter_mask(mask, min_component):
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    out = np.zeros_like(mask)
    for i in range(1, num_labels):
        if stats[i, cv2.CC_STAT_AREA] >= min_component:
            out[labels == i] = 1
    return out


def postprocess(probs, thresholds, min_component, min_pixels):
    B, C, H, W = probs.shape
    out = np.zeros((B, C, H, W), dtype=np.uint8)
    for c in range(C):
        base = (probs[:, c].astype(np.float32) >= thresholds[c]).astype(np.uint8)
        if min_component[c] > 0 and base.any():
            for b in range(B):
                base[b] = cc_filter_mask(base[b], min_component[c])
        for b in range(B):
            if base[b].sum() >= min_pixels[c]:
                out[b, c] = base[b]
    return out
```

```python
# ===== stage 1/3 | resolve the 5 checkpoints and load the test images =====
CKPTS = {}
for p in sorted(Path("/kaggle/input").rglob("*.pth")):
    if "FTB" not in p.name:
        tag = "baseline"
    else:
        fold_num = p.name.split("FOLD")[1].split("FTB")[0]
        tag = f"ft{fold_num}b"
    CKPTS[tag] = p
print("checkpoints:", {k: str(v) for k, v in CKPTS.items()}, flush=True)
assert set(CKPTS) == {"baseline", "ft1b", "ft3b", "ft4b", "ft5b"}, f"unexpected members {sorted(CKPTS)}"

sub_raw = pd.read_csv(SAMPLE)
test_ids = sub_raw["image_id"].unique()
H, W = 128, 800
test_u8 = np.empty((len(test_ids), H, W), dtype=np.uint8)
for i, iid in enumerate(tqdm(test_ids, desc="test imgs", dynamic_ncols=True)):
    name = iid if Path(iid).suffix else f"{iid}.jpg"
    with Image.open(TEST_DIR / name) as im:
        test_u8[i] = np.asarray(im.convert("L"), dtype=np.uint8)
print(f"test_u8 {test_u8.nbytes/1e9:.2f}GB | {len(test_ids)} imgs | {time.time()-t_start:.0f}s", flush=True)
```

```python
# ===== stage 2/3 | hflip-TTA inference for every blend member, uint8 dumps =====
@torch.inference_mode()
def probs_tta(model, u8, tag=""):
    out = np.empty((len(u8), 4, H, W), dtype=np.float16)
    t0 = time.time()
    for s in range(0, len(u8), BATCH):
        x = torch.from_numpy(u8[s:s + BATCH]).unsqueeze(1).float().div_(255).to(device, non_blocking=True)
        with torch.amp.autocast("cuda", dtype=torch.float16):
            logits = model(x)
            logits = (logits + torch.flip(model(torch.flip(x, dims=[-1])), dims=[-1])) / 2
        out[s:s + BATCH] = torch.sigmoid(logits).float().cpu().numpy().astype(np.float16)
        done = min(s + BATCH, len(u8))
        if (s // BATCH) % 20 == 0:
            el = time.time() - t0
            print(f"  {tag} {done}/{len(u8)} | {el:.0f}s | ETA {el/done*(len(u8)-done):.0f}s", flush=True)
    return out


for i, (tag, ckpt) in enumerate(sorted(CKPTS.items()), 1):
    model = torch.load(ckpt, map_location=device, weights_only=False).to(device).eval()
    print(f"\n[{i}/5] {tag} | {sum(p.numel() for p in model.parameters())/1e6:.2f}M params | "
          f"{time.time()-t_start:.0f}s elapsed", flush=True)
    tp = probs_tta(model, test_u8, tag=tag)
    u8 = np.clip(tp * 255.0, 0, 255).round().astype(np.uint8)
    np.save(f"/kaggle/working/test_probs_{tag}.npy", u8)
    print(f"[{i}/5] {tag} saved uint8 | mean {float(tp.mean()):.4f} | {time.time()-t_start:.0f}s", flush=True)
    del model, tp, u8
    gc.collect(); torch.cuda.empty_cache()
```

## 3/3 - blend and calibrated masks


```python

import time, json
import numpy as np, pandas as pd, cv2
from pathlib import Path

t_start = time.time()
SETTINGS = {"weights": {"baseline": 1.0, "ft1b": 1.0, "ft3b": 1.0, "ft4b": 1.0, "ft5b": 1.0}, "mode": "mean", "target_masks": {"defect_a": 670, "defect_b": 492, "defect_c": 1468, "defect_d": 1180}, "min_component": [48, 32, 160, 256], "min_pixels": [64, 0, 192, 384]}
WEIGHTS = {k: float(v) for k, v in SETTINGS["weights"].items() if float(v) > 0}
MODE = SETTINGS.get("mode", "mean")  # "mean" = raw prob average, "zscore" = per-model/class standardised
TARGET = SETTINGS["target_masks"]
MC, MT = SETTINGS["min_component"], SETTINGS["min_pixels"]
BATCH = 256
DEFECT_TYPES = ["defect_a", "defect_b", "defect_c", "defect_d"]
ROOT = Path("/kaggle/input/competitions/industrial-surface-defect-detection/public/")
if not (ROOT / "sample_submission.csv").exists():  # Kaggle mounts the competition data at varying depths
    _hits = sorted(Path("/kaggle/input").rglob("sample_submission.csv"))
    if _hits:
        ROOT = _hits[0].parent
print("ROOT:", ROOT, flush=True)
assert (ROOT / "sample_submission.csv").exists(), f"competition data not mounted: {ROOT}"
SAMPLE = ROOT / "sample_submission.csv"
print("SETTINGS:", json.dumps(SETTINGS), flush=True)


def rle_encode(mask):
    pixels = (np.asarray(mask) > 0).flatten(order="F")
    padded = np.concatenate([[0], pixels, [0]])
    changes = np.flatnonzero(padded[1:] != padded[:-1]) + 1
    changes[1::2] -= changes[::2]
    return " ".join(map(str, changes))


def cc_filter_mask(mask, min_component):
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    out = np.zeros_like(mask)
    for i in range(1, num_labels):
        if stats[i, cv2.CC_STAT_AREA] >= min_component:
            out[labels == i] = 1
    return out


def class_masks(probs_c, thr, min_component, min_pixels):
    base = (probs_c >= thr).astype(np.uint8)
    out = np.zeros_like(base)
    for b in range(base.shape[0]):
        m = cc_filter_mask(base[b], min_component) if (min_component > 0 and base[b].any()) else base[b]
        if m.sum() >= min_pixels:
            out[b] = m
    return out


sub_raw = pd.read_csv(SAMPLE)
test_ids = sub_raw["image_id"].unique()

found = {}
for p in [q for root in ("/kaggle/input", "/kaggle/working") for q in Path(root).rglob("test_probs_*.npy")]:
    tag = p.name[len("test_probs_"):-len(".npy")]
    tag = tag[len("fold"):] if tag.startswith("fold") else tag  # fold1 -> 1
    found[tag] = p
print("prob files:", {k: str(v.parent) for k, v in sorted(found.items())}, flush=True)
missing = [k for k in WEIGHTS if k not in found]
assert not missing, f"missing {missing}; found {sorted(found)}"
probs = {k: np.load(found[k], mmap_mode="r") for k in WEIGHTS}
Q8 = {k: probs[k].dtype == np.uint8 for k in WEIGHTS}  # uint8 prob dumps are dequantised on the fly
n_img, C, H, W = next(iter(probs.values())).shape
wsum = float(sum(WEIGHTS.values()))
print(f"probs {n_img} x {C} x {H}x{W} | weights {WEIGHTS} | {time.time()-t_start:.0f}s", flush=True)


STATS = None
if MODE == "zscore":
    n_px = 0
    s1 = {k: np.zeros(C, np.float64) for k in WEIGHTS}
    s2 = {k: np.zeros(C, np.float64) for k in WEIGHTS}
    for i0 in range(0, n_img, BATCH):
        for k in WEIGHTS:
            piece = np.asarray(probs[k][i0:i0 + BATCH], dtype=np.float32) / (255.0 if Q8[k] else 1.0)
            s1[k] += piece.sum(axis=(0, 2, 3))
            s2[k] += np.square(piece).sum(axis=(0, 2, 3))
        n_px += piece.shape[0] * H * W
    STATS = {k: (s1[k] / n_px, np.sqrt(np.maximum(s2[k] / n_px - (s1[k] / n_px) ** 2, 1e-8))) for k in WEIGHTS}
    print("model stats (mean/std per class):", {k: (np.round(m, 4).tolist(), np.round(d, 4).tolist()) for k, (m, d) in STATS.items()}, flush=True)


def blend(i0, i1):
    acc = None
    for k, w in WEIGHTS.items():
        piece = np.asarray(probs[k][i0:i1], dtype=np.float32) / (255.0 if Q8[k] else 1.0)
        if MODE == "zscore":
            mu, sd = STATS[k]
            piece = (piece - mu.reshape(1, C, 1, 1)) / sd.reshape(1, C, 1, 1)
        acc = piece * w if acc is None else acc + piece * w
    return acc / wsum


mx = np.zeros((C, n_img), dtype=np.float32)
for i0 in range(0, n_img, BATCH):
    mx[:, i0:i0 + BATCH] = blend(i0, i0 + BATCH).max(axis=(2, 3)).T
print(f"per-image max probs done | {time.time()-t_start:.0f}s", flush=True)


def count_class(c, thr):
    n = 0
    for i0 in range(0, n_img, BATCH):
        n += int((class_masks(blend(i0, i0 + BATCH)[:, c], thr, MC[c], MT[c]).sum(axis=(1, 2)) > 0).sum())
    return n


thresholds = []
for c, dt in enumerate(DEFECT_TYPES):
    target = int(TARGET[dt])
    order = np.sort(mx[c])[::-1]
    thr = float(order[min(max(target - 1, 0), len(order) - 1)]) - 1e-4
    for attempt in range(4):
        cnt = count_class(c, thr)
        print(f"  {dt} try{attempt}: thr={thr:.4f} masks={cnt} target={target} | {time.time()-t_start:.0f}s", flush=True)
        if cnt >= target or attempt == 3:
            break
        need = target - cnt
        idx = min(target + need - 1, len(order) - 1)
        thr = float(order[max(idx - 1, 0)]) - 1e-4
    thresholds.append(round(thr, 4))
print(f"CALIBRATED thresholds {thresholds} | {time.time()-t_start:.0f}s", flush=True)

rle_dict = {}
for i0 in range(0, n_img, BATCH):
    bl = blend(i0, i0 + BATCH)
    for c, dt in enumerate(DEFECT_TYPES):
        m = class_masks(bl[:, c], thresholds[c], MC[c], MT[c])
        for b in range(m.shape[0]):
            if m[b].sum() > 0:
                rle_dict[(str(test_ids[i0 + b]), dt)] = rle_encode(m[b])
    if (i0 // BATCH) % 4 == 0:
        print(f"  postproc {min(i0+BATCH, n_img)}/{n_img} | {time.time()-t_start:.0f}s", flush=True)

sub = sub_raw.copy()
sub["mask_rle"] = [rle_dict.get((str(r.image_id), str(r.defect_type)), " ") for r in sub.itertuples(index=False)]
sub = sub[list(sub_raw.columns)]
sub.to_csv("/kaggle/working/submission.csv", index=False)
non_empty = int((sub["mask_rle"] != " ").sum())
print(f"\nsubmission.csv | rows {len(sub)} | masks {non_empty} ({non_empty/len(sub):.2%})", flush=True)
print("per class masks:", sub[sub.mask_rle != " "].defect_type.value_counts().reindex(DEFECT_TYPES).to_dict(), flush=True)

chk = pd.read_csv("/kaggle/working/submission.csv")
assert list(chk.columns) == list(sub_raw.columns)
assert len(chk) == len(sub_raw)
assert set(chk["image_id"]) == set(sub_raw["image_id"])
assert chk["mask_rle"].notna().all()
assert chk["mask_rle"].str.match(r"^(\s*|\d+( \d+)*)$").all()
print(f"VALIDATION PASSED | {time.time()-t_start:.0f}s", flush=True)
```
